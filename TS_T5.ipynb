Simplify sentences with the TS-T5 models from the paper. Models: `kcsheang/ts-t5-base-4tokens`, `kcsheang/ts-t5-base-5tokens`.

```python
from ts_t5.simplify import Simplifier

simplifier = Simplifier('kcsheang/ts-t5-base-4tokens')
```

```python
sentences = [
    'The incumbent was defeated in the election by a substantial margin .',
    'The cat perched precariously on the windowsill , surveying the courtyard below .',
]
for s in sentences:
    print(s)
    print('->', simplifier.simplify(s), '\n')
```
